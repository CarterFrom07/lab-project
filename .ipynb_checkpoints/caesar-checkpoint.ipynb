```python
name = input("What is your name?")
name = name.casefold()
name = list(name)

key = input("What is the key? (number from 1-25)")
key = int(key)

ciphertext = []
for letter in name:
    idx = ord(letter) - ord('a')
    if idx >= 26:
        ciphertext.append(letter)
        continue
    idx += key
    idx %= 26
    ciphertext.append(chr(ord('A')+idx))

ciphertext = "".join(ciphertext)
print(f"Your encrypted name is {ciphertext}. (To unencrypt, input the encrypted name with and use the key 26-k, where k is the encryption key)")
```
